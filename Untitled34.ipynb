```python
"""
=============================================================
  CROP YIELD PREDICTION — TRANSFORMER MODEL (TABULAR DATA)
=============================================================
Dataset : merged_crop_dataset.xls  (19199 rows × 25 columns)
Target  : yield
Model   : TabTransformer / FT-Transformer style
          (categorical embedding + numeric MLP + Transformer encoder)
=============================================================
"""

# ── 0. INSTALL DEPENDENCIES ───────────────────────────────────────────────────
# Run this once before executing the script:
#   pip install torch scikit-learn pandas numpy matplotlib seaborn

# ── 1. IMPORTS ────────────────────────────────────────────────────────────────
import math
import warnings
import numpy  as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection  import train_test_split
from sklearn.preprocessing    import LabelEncoder, RobustScaler
from sklearn.metrics          import mean_absolute_error, mean_squared_error, r2_score

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

warnings.filterwarnings("ignore")
torch.manual_seed(42)
np.random.seed(42)

# ── 2. CONFIGURATION ──────────────────────────────────────────────────────────
DATA_PATH   = "merged_crop_dataset.xls"   # ← change if file is elsewhere
TARGET      = "yield"

# Transformer hyper-parameters
D_MODEL     = 64      # embedding / model dimension
N_HEADS     = 4       # attention heads
N_LAYERS    = 3       # transformer encoder layers
DIM_FF      = 128     # feed-forward hidden dim
DROPOUT     = 0.1
BATCH_SIZE  = 256
EPOCHS      = 60
LR          = 1e-3
PATIENCE    = 10      # early-stopping patience
DEVICE      = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Using device: {DEVICE}")

# ── 3. LOAD DATA ──────────────────────────────────────────────────────────────
print("\n[1/6] Loading data …")
df = pd.read_csv(r"C:\Users\chinna\OneDrive\Desktop\AI IN AGRICULTURE\transformer datset\meged_dataset.csv")
print(f"  Shape : {df.shape}")
print(f"  Columns: {df.columns.tolist()}")
print(f"  Missing values: {df.isnull().sum().sum()}")

# ── 4. FEATURE ENGINEERING ────────────────────────────────────────────────────
print("\n[2/6] Feature engineering …")

# Remove extreme outliers in target (top 1%)
upper = df[TARGET].quantile(0.99)
df    = df[df[TARGET] <= upper].reset_index(drop=True)
print(f"  Rows after outlier removal: {len(df)}")

# Log-transform the heavily skewed target
df["log_yield"] = np.log1p(df[TARGET])

# Categorical & numeric columns
CAT_COLS = ["crop", "season", "state"]
NUM_COLS = [c for c in df.columns
            if c not in CAT_COLS + [TARGET, "log_yield", "year"]]

# Add year as numeric feature
NUM_COLS = ["year"] + NUM_COLS

print(f"  Categorical features : {CAT_COLS}")
print(f"  Numeric features     : {NUM_COLS}")

# Encode categoricals
label_encoders = {}
for col in CAT_COLS:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    label_encoders[col] = le

# Category cardinalities (needed for embedding layers)
cat_cardinalities = [df[col].nunique() for col in CAT_COLS]
print(f"  Category cardinalities: {dict(zip(CAT_COLS, cat_cardinalities))}")

# Scale numeric features
scaler = RobustScaler()
df[NUM_COLS] = scaler.fit_transform(df[NUM_COLS])

# ── 5. TRAIN / VAL / TEST SPLIT ───────────────────────────────────────────────
print("\n[3/6] Splitting dataset …")
X_cat = df[CAT_COLS].values.astype(np.int64)
X_num = df[NUM_COLS].values.astype(np.float32)
y     = df["log_yield"].values.astype(np.float32)

X_cat_tr, X_cat_tmp, X_num_tr, X_num_tmp, y_tr, y_tmp = train_test_split(
    X_cat, X_num, y, test_size=0.2, random_state=42)
X_cat_val, X_cat_te, X_num_val, X_num_te, y_val, y_te = train_test_split(
    X_cat_tmp, X_num_tmp, y_tmp, test_size=0.5, random_state=42)

print(f"  Train : {len(y_tr):,}  |  Val : {len(y_val):,}  |  Test : {len(y_te):,}")

# ── 6. PYTORCH DATASET ────────────────────────────────────────────────────────
class CropDataset(Dataset):
    def __init__(self, cat, num, target):
        self.cat    = torch.tensor(cat,    dtype=torch.long)
        self.num    = torch.tensor(num,    dtype=torch.float32)
        self.target = torch.tensor(target, dtype=torch.float32)

    def __len__(self):
        return len(self.target)

    def __getitem__(self, idx):
        return self.cat[idx], self.num[idx], self.target[idx]

train_ds = CropDataset(X_cat_tr,  X_num_tr,  y_tr)
val_ds   = CropDataset(X_cat_val, X_num_val, y_val)
test_ds  = CropDataset(X_cat_te,  X_num_te,  y_te)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
val_dl   = DataLoader(val_ds,   batch_size=BATCH_SIZE)
test_dl  = DataLoader(test_ds,  batch_size=BATCH_SIZE)

# ── 7. MODEL: FT-TRANSFORMER ──────────────────────────────────────────────────
"""
Architecture (Feature Tokenizer + Transformer):

  Categorical cols  →  Embedding(cardinality, d_model)  ┐
  Numeric cols      →  Linear(1, d_model)  (per-feature) ┘  → [seq_len × d_model]
                                                              → Transformer Encoder
                                                              → Mean-pool
                                                              → MLP head
                                                              → scalar output
"""

class FeatureTokenizer(nn.Module):
    """Embed each feature (cat or num) into d_model-dim token."""
    def __init__(self, cat_cardinalities, n_num, d_model):
        super().__init__()
        # One embedding table per categorical feature
        self.cat_embeddings = nn.ModuleList([
            nn.Embedding(card + 1, d_model) for card in cat_cardinalities
        ])
        # One linear per numeric feature  (weight shape: d_model × 1)
        self.num_projections = nn.ModuleList([
            nn.Linear(1, d_model) for _ in range(n_num)
        ])

    def forward(self, cat, num):
        # cat : (B, n_cat)   num : (B, n_num)
        tokens = []
        for i, emb in enumerate(self.cat_embeddings):
            tokens.append(emb(cat[:, i]))               # (B, d_model)
        for i, proj in enumerate(self.num_projections):
            tokens.append(proj(num[:, i:i+1]))           # (B, d_model)
        return torch.stack(tokens, dim=1)                # (B, seq_len, d_model)


class CropTransformer(nn.Module):
    def __init__(self, cat_cardinalities, n_num, d_model, n_heads,
                 n_layers, dim_ff, dropout):
        super().__init__()
        self.tokenizer = FeatureTokenizer(cat_cardinalities, n_num, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model        = d_model,
            nhead          = n_heads,
            dim_feedforward= dim_ff,
            dropout        = dropout,
            batch_first    = True,    # (B, seq, d_model)
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        # Regression head
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model // 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_model // 2, 1),
        )

    def forward(self, cat, num):
        x = self.tokenizer(cat, num)      # (B, seq_len, d_model)
        x = self.transformer(x)           # (B, seq_len, d_model)
        x = x.mean(dim=1)                 # (B, d_model)  — mean pooling
        return self.head(x).squeeze(-1)   # (B,)


model = CropTransformer(
    cat_cardinalities = cat_cardinalities,
    n_num             = len(NUM_COLS),
    d_model           = D_MODEL,
    n_heads           = N_HEADS,
    n_layers          = N_LAYERS,
    dim_ff            = DIM_FF,
    dropout           = DROPOUT,
).to(DEVICE)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\n[4/6] Model built — trainable parameters: {total_params:,}")
print(model)

# ── 8. TRAINING LOOP ──────────────────────────────────────────────────────────
print("\n[5/6] Training …")

criterion = nn.HuberLoss()     # robust to remaining outliers
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

train_losses, val_losses = [], []
best_val_loss = float("inf")
patience_ctr  = 0

for epoch in range(1, EPOCHS + 1):
    # — train —
    model.train()
    t_loss = 0.0
    for cat_b, num_b, y_b in train_dl:
        cat_b, num_b, y_b = cat_b.to(DEVICE), num_b.to(DEVICE), y_b.to(DEVICE)
        optimizer.zero_grad()
        pred = model(cat_b, num_b)
        loss = criterion(pred, y_b)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        t_loss += loss.item() * len(y_b)
    t_loss /= len(train_ds)
    scheduler.step()

    # — validate —
    model.eval()
    v_loss = 0.0
    with torch.no_grad():
        for cat_b, num_b, y_b in val_dl:
            cat_b, num_b, y_b = cat_b.to(DEVICE), num_b.to(DEVICE), y_b.to(DEVICE)
            pred   = model(cat_b, num_b)
            v_loss += criterion(pred, y_b).item() * len(y_b)
    v_loss /= len(val_ds)

    train_losses.append(t_loss)
    val_losses.append(v_loss)

    if epoch % 5 == 0 or epoch == 1:
        print(f"  Epoch {epoch:3d}/{EPOCHS}  "
              f"train_loss={t_loss:.4f}  val_loss={v_loss:.4f}")

    # — early stopping —
    if v_loss < best_val_loss:
        best_val_loss = v_loss
        torch.save(model.state_dict(), "best_crop_transformer.pt")
        patience_ctr = 0
    else:
        patience_ctr += 1
        if patience_ctr >= PATIENCE:
            print(f"  Early stopping at epoch {epoch}")
            break

# ── 9. EVALUATION ─────────────────────────────────────────────────────────────
print("\n[6/6] Evaluating on test set …")
model.load_state_dict(torch.load("best_crop_transformer.pt", map_location=DEVICE))
model.eval()

all_preds, all_true = [], []
with torch.no_grad():
    for cat_b, num_b, y_b in test_dl:
        cat_b, num_b = cat_b.to(DEVICE), num_b.to(DEVICE)
        pred = model(cat_b, num_b).cpu().numpy()
        all_preds.extend(pred)
        all_true.extend(y_b.numpy())

# Back-transform from log-space
y_pred_log = np.array(all_preds)
y_true_log = np.array(all_true)
y_pred_orig = np.expm1(y_pred_log)
y_true_orig = np.expm1(y_true_log)

mae  = mean_absolute_error(y_true_orig, y_pred_orig)
rmse = np.sqrt(mean_squared_error(y_true_orig, y_pred_orig))
r2   = r2_score(y_true_orig, y_pred_orig)

print(f"\n  ╔══════════════════════════════╗")
print(f"  ║  TEST SET RESULTS            ║")
print(f"  ╠══════════════════════════════╣")
print(f"  ║  MAE  : {mae:>10.4f}          ║")
print(f"  ║  RMSE : {rmse:>10.4f}          ║")
print(f"  ║  R²   : {r2:>10.4f}          ║")
print(f"  ╚══════════════════════════════╝")

# ── 10. PLOTS ─────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Crop Yield — FT-Transformer Results", fontsize=14, fontweight="bold")

# Loss curves
ax = axes[0]
ax.plot(train_losses, label="Train Loss", color="steelblue")
ax.plot(val_losses,   label="Val Loss",   color="tomato")
ax.set_title("Training & Validation Loss")
ax.set_xlabel("Epoch")
ax.set_ylabel("Huber Loss")
ax.legend()
ax.grid(alpha=0.3)

# Actual vs Predicted (clip for visibility)
ax = axes[1]
clip = np.percentile(y_true_orig, 99)
mask = y_true_orig <= clip
ax.scatter(y_true_orig[mask], y_pred_orig[mask], alpha=0.3, s=10, color="steelblue")
lim  = max(y_true_orig[mask].max(), y_pred_orig[mask].max())
ax.plot([0, lim], [0, lim], "r--", lw=1.5, label="Perfect")
ax.set_title(f"Actual vs Predicted (R²={r2:.3f})")
ax.set_xlabel("Actual Yield")
ax.set_ylabel("Predicted Yield")
ax.legend()
ax.grid(alpha=0.3)

# Residuals distribution
ax = axes[2]
residuals = y_true_orig - y_pred_orig
residuals_clipped = np.clip(residuals,
                            np.percentile(residuals, 1),
                            np.percentile(residuals, 99))
ax.hist(residuals_clipped, bins=60, color="steelblue", edgecolor="white", alpha=0.8)
ax.axvline(0, color="tomato", lw=2, linestyle="--")
ax.set_title("Residuals Distribution")
ax.set_xlabel("Actual − Predicted")
ax.set_ylabel("Count")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("transformer_results.png", dpi=150, bbox_inches="tight")
plt.show()
print("\n  Plot saved → transformer_results.png")
print("  Model checkpoint saved → best_crop_transformer.pt")
print("\nDone! ✓")
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

DEVICE     = "cuda" if torch.cuda.is_available() else "cpu"
D_MODEL    = 64
N_HEADS    = 4
N_LAYERS   = 3
DIM_FF     = 128
DROPOUT    = 0.1
BATCH_SIZE = 256
EPOCHS     = 100
LR         = 1e-3
PATIENCE   = 15

class CropDataset(Dataset):
    def __init__(self, cat, num, target):
        self.cat    = torch.tensor(cat,    dtype=torch.long)
        self.num    = torch.tensor(num,    dtype=torch.float32)
        self.target = torch.tensor(target, dtype=torch.float32)
    def __len__(self): return len(self.target)
    def __getitem__(self, i): return self.cat[i], self.num[i], self.target[i]

train_dl = DataLoader(CropDataset(X_cat_tr,  X_num_tr,  y_tr),
                      batch_size=BATCH_SIZE, shuffle=True)
val_dl   = DataLoader(CropDataset(X_cat_val, X_num_val, y_val),
                      batch_size=BATCH_SIZE)
test_dl  = DataLoader(CropDataset(X_cat_te,  X_num_te,  y_te),
                      batch_size=BATCH_SIZE)

class FeatureTokenizer(nn.Module):
    def __init__(self, cat_cardinalities, n_num, d_model):
        super().__init__()
        self.cat_embs  = nn.ModuleList(
            [nn.Embedding(c+1, d_model) for c in cat_cardinalities])
        self.num_projs = nn.ModuleList(
            [nn.Linear(1, d_model) for _ in range(n_num)])
    def forward(self, cat, num):
        tokens  = [emb(cat[:,i]) for i,emb in enumerate(self.cat_embs)]
        tokens += [proj(num[:,i:i+1]) for i,proj in enumerate(self.num_projs)]
        return torch.stack(tokens, dim=1)

class CropTransformer(nn.Module):
    def __init__(self, cat_cardinalities, n_num, d_model,
                 n_heads, n_layers, dim_ff, dropout):
        super().__init__()
        self.tokenizer   = FeatureTokenizer(
            cat_cardinalities, n_num, d_model)
        enc_layer        = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=n_heads,
            dim_feedforward=dim_ff, dropout=dropout,
            batch_first=True)
        self.transformer = nn.TransformerEncoder(
            enc_layer, num_layers=n_layers)
        self.head        = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model//2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_model//2, 1))
    def forward(self, cat, num):
        x = self.tokenizer(cat, num)
        x = self.transformer(x)
        x = x.mean(dim=1)
        return self.head(x).squeeze(-1)

# ── Build V1 model ──
model_v1 = CropTransformer(
    cat_cardinalities=cat_cardinalities,
    n_num=len(NUM_COLS),
    d_model=D_MODEL,
    n_heads=N_HEADS,
    n_layers=N_LAYERS,
    dim_ff=DIM_FF,
    dropout=DROPOUT).to(DEVICE)

n_params = sum(p.numel() for p in model_v1.parameters())
print(f"V1 Parameters: {n_params:,}")

# ── Train ──
criterion = nn.HuberLoss()
optimizer = torch.optim.AdamW(
    model_v1.parameters(), lr=LR, weight_decay=1e-2)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=EPOCHS)

best_val  = float("inf")
pat_count = 0

print(f"\nTraining V1 on temporal split...")
print(f"{'Epoch':>6} | {'Train Loss':>10} | {'Val Loss':>10}")
print("-" * 35)

for epoch in range(1, EPOCHS+1):
    # Train
    model_v1.train()
    t_loss = 0.0
    for cat_b, num_b, y_b in train_dl:
        cat_b, num_b, y_b = (cat_b.to(DEVICE), 
                              num_b.to(DEVICE), 
                              y_b.to(DEVICE))
        optimizer.zero_grad()
        loss = criterion(model_v1(cat_b, num_b), y_b)
        loss.backward()
        nn.utils.clip_grad_norm_(model_v1.parameters(), 1.0)
        optimizer.step()
        t_loss += loss.item() * len(y_b)
    t_loss /= len(train_dl.dataset)
    scheduler.step()

    # Validate
    model_v1.eval()
    v_loss = 0.0
    with torch.no_grad():
        for cat_b, num_b, y_b in val_dl:
            cat_b, num_b, y_b = (cat_b.to(DEVICE),
                                  num_b.to(DEVICE),
                                  y_b.to(DEVICE))
            v_loss += criterion(
                model_v1(cat_b, num_b), y_b).item() * len(y_b)
    v_loss /= len(val_dl.dataset)

    if epoch % 10 == 0 or epoch == 1:
        print(f"{epoch:>6} | {t_loss:>10.5f} | {v_loss:>10.5f}")

    if v_loss < best_val:
        best_val  = v_loss
        pat_count = 0
        torch.save(model_v1.state_dict(), 
                   "best_v1_temporal.pt")
    else:
        pat_count += 1
        if pat_count >= PATIENCE:
            print(f"Early stopping at epoch {epoch}")
            break

print(f"\nBest val loss: {best_val:.5f}")

# ── Evaluate ──
model_v1.load_state_dict(
    torch.load("best_v1_temporal.pt", map_location=DEVICE))
model_v1.eval()

preds, trues = [], []
with torch.no_grad():
    for cat_b, num_b, y_b in test_dl:
        cat_b, num_b = cat_b.to(DEVICE), num_b.to(DEVICE)
        preds.extend(model_v1(cat_b, num_b).cpu().numpy())
        trues.extend(y_b.numpy())

y_pred_v1 = np.expm1(np.array(preds))
y_true_v1 = np.expm1(np.array(trues))

mae  = mean_absolute_error(y_true_v1, y_pred_v1)
rmse = np.sqrt(mean_squared_error(y_true_v1, y_pred_v1))
r2   = r2_score(y_true_v1, y_pred_v1)

print(f"\n{'='*35}")
print(f"V1 TEMPORAL SPLIT RESULTS")
print(f"{'='*35}")
print(f"MAE  = {mae:.4f}")
print(f"RMSE = {rmse:.4f}")
print(f"R²   = {r2:.4f}")
print(f"{'='*35}")
```

```python
# ── Simple TabTransformer Implementation ──
import torch
import torch.nn as nn
from torch.utils.data import DataLoader

class TabTransformer(nn.Module):
    def __init__(self, cat_cardinalities, n_num, 
                 d_model=32, n_heads=4, n_layers=3, dropout=0.1):
        super().__init__()
        # Categorical embeddings with transformer
        self.cat_embs = nn.ModuleList([
            nn.Embedding(c+1, d_model) 
            for c in cat_cardinalities])
        
        enc_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=n_heads,
            dim_feedforward=d_model*4,
            dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(
            enc_layer, num_layers=n_layers)
        
        # MLP for numerical features
        self.num_mlp = nn.Sequential(
            nn.Linear(n_num, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 32))
        
        # Final head
        cat_out_dim = len(cat_cardinalities) * d_model
        self.head = nn.Sequential(
            nn.Linear(cat_out_dim + 32, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1))
    
    def forward(self, cat, num):
        # Process categoricals through transformer
        cat_tokens = torch.stack([
            emb(cat[:,i]) 
            for i,emb in enumerate(self.cat_embs)], dim=1)
        cat_out = self.transformer(cat_tokens)
        cat_out = cat_out.flatten(1)
        
        # Process numericals through MLP
        num_out = self.num_mlp(num)
        
        # Concatenate and predict
        out = torch.cat([cat_out, num_out], dim=1)
        return self.head(out).squeeze(-1)

# ── Train TabTransformer ──
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

tab_model = TabTransformer(
    cat_cardinalities=cat_cardinalities,
    n_num=len(NUM_COLS),
    d_model=32,
    n_heads=4,
    n_layers=3,
    dropout=0.1).to(DEVICE)

n_params_tab = sum(p.numel() for p in tab_model.parameters())
print(f"TabTransformer Parameters: {n_params_tab:,}")

# DataLoaders
train_dl_tab = DataLoader(
    CropDataset(X_cat_tr,  X_num_tr,  y_tr),
    batch_size=256, shuffle=True)
val_dl_tab   = DataLoader(
    CropDataset(X_cat_val, X_num_val, y_val),
    batch_size=256)
test_dl_tab  = DataLoader(
    CropDataset(X_cat_te,  X_num_te,  y_te),
    batch_size=256)

criterion_tab = nn.HuberLoss()
optimizer_tab = torch.optim.AdamW(
    tab_model.parameters(), lr=1e-3, weight_decay=1e-2)
scheduler_tab = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer_tab, T_max=60)

best_val_tab  = float("inf")
pat_count_tab = 0
PATIENCE_TAB  = 15

print(f"\nTraining TabTransformer...")
print(f"{'Epoch':>6} | {'Train Loss':>10} | {'Val Loss':>10}")
print("-" * 35)

for epoch in range(1, 61):
    # Train
    tab_model.train()
    t_loss = 0.0
    for cat_b, num_b, y_b in train_dl_tab:
        cat_b, num_b, y_b = (cat_b.to(DEVICE),
                              num_b.to(DEVICE),
                              y_b.to(DEVICE))
        optimizer_tab.zero_grad()
        loss = criterion_tab(tab_model(cat_b, num_b), y_b)
        loss.backward()
        nn.utils.clip_grad_norm_(tab_model.parameters(), 1.0)
        optimizer_tab.step()
        t_loss += loss.item() * len(y_b)
    t_loss /= len(train_dl_tab.dataset)
    scheduler_tab.step()

    # Validate
    tab_model.eval()
    v_loss = 0.0
    with torch.no_grad():
        for cat_b, num_b, y_b in val_dl_tab:
            cat_b, num_b, y_b = (cat_b.to(DEVICE),
                                  num_b.to(DEVICE),
                                  y_b.to(DEVICE))
            v_loss += criterion_tab(
                tab_model(cat_b, num_b), y_b).item() * len(y_b)
    v_loss /= len(val_dl_tab.dataset)

    if epoch % 10 == 0 or epoch == 1:
        print(f"{epoch:>6} | {t_loss:>10.5f} | {v_loss:>10.5f}")

    if v_loss < best_val_tab:
        best_val_tab  = v_loss
        pat_count_tab = 0
        torch.save(tab_model.state_dict(), "best_tab_temporal.pt")
    else:
        pat_count_tab += 1
        if pat_count_tab >= PATIENCE_TAB:
            print(f"Early stopping at epoch {epoch}")
            break

print(f"\nBest val loss: {best_val_tab:.5f}")

# Evaluate
tab_model.load_state_dict(
    torch.load("best_tab_temporal.pt", map_location=DEVICE))
tab_model.eval()

preds_tab, trues_tab = [], []
with torch.no_grad():
    for cat_b, num_b, y_b in test_dl_tab:
        cat_b, num_b = cat_b.to(DEVICE), num_b.to(DEVICE)
        preds_tab.extend(tab_model(cat_b, num_b).cpu().numpy())
        trues_tab.extend(y_b.numpy())

y_pred_tab = np.expm1(np.array(preds_tab))
y_true_tab = np.expm1(np.array(trues_tab))

mae_tab  = mean_absolute_error(y_true_tab, y_pred_tab)
rmse_tab = np.sqrt(mean_squared_error(y_true_tab, y_pred_tab))
r2_tab   = r2_score(y_true_tab, y_pred_tab)

print(f"\n{'='*40}")
print(f"COMPLETE RESULTS — ALL MODELS")
print(f"{'='*40}")
print(f"{'Model':<20} {'MAE':>8} {'RMSE':>8} {'R²':>8}")
print("-" * 50)
print(f"{'Random Forest':<20} {'0.7958':>8} {'3.3846':>8} {'0.9010':>8}")
print(f"{'XGBoost':<20} {'0.6634':>8} {'2.6766':>8} {'0.9381':>8}")
print(f"{'TabTransformer':<20} {mae_tab:>8.4f} {rmse_tab:>8.4f} {r2_tab:>8.4f}")
print(f"{'FT-Trans V1':<20} {'0.7750':>8} {'2.4043':>8} {'0.9501':>8}")
print(f"{'FT-Trans V2/FAST':<20} {'0.9944':>8} {'3.1214':>8} {'0.9158':>8}")
print(f"{'='*50}")
```

```python
import numpy as np
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

def bootstrap_ci(y_true, y_pred, n_iter=1000, ci=95):
    mae_scores, rmse_scores, r2_scores = [], [], []
    n = len(y_true)
    rng = np.random.RandomState(42)
    
    for _ in range(n_iter):
        idx = rng.randint(0, n, n)
        yt  = y_true[idx]
        yp  = y_pred[idx]
        
        mae_scores.append(mean_absolute_error(yt, yp))
        rmse_scores.append(np.sqrt(mean_squared_error(yt, yp)))
        r2_scores.append(r2_score(yt, yp))
    
    alpha = (100 - ci) / 2
    return {
        'MAE':  (np.percentile(mae_scores,  alpha),
                 np.percentile(mae_scores,  100-alpha)),
        'RMSE': (np.percentile(rmse_scores, alpha),
                 np.percentile(rmse_scores, 100-alpha)),
        'R2':   (np.percentile(r2_scores,   alpha),
                 np.percentile(r2_scores,   100-alpha))
    }

# Run for all models
print("Computing Bootstrap CIs (1000 iterations)...")
print("="*75)

models = {
    'Random Forest':     (y_true_v1,  y_pred_rf),
    'XGBoost':           (y_true_v1,  y_pred_xgb),
    'TabTransformer':    (y_true_tab, y_pred_tab),
    'FT-Trans V1':       (y_true_v1,  y_pred_v1),
    'FT-Trans V2/FAST':  (y_true_v2,  y_pred_v2),
}

print(f"{'Model':<20} {'MAE [95% CI]':>22} {'RMSE [95% CI]':>22} {'R² [95% CI]':>22}")
print("-"*75)

for name, (yt, yp) in models.items():
    ci = bootstrap_ci(yt, yp)
    print(f"{name:<20} "
          f"{f'[{ci['MAE'][0]:.3f}–{ci['MAE'][1]:.3f}]':>22}"
          f"{f'[{ci['RMSE'][0]:.3f}–{ci['RMSE'][1]:.3f}]':>22} "
          f"{f'[{ci['R2'][0]:.3f}–{ci['R2'][1]:.3f}]':>22}")

print("="*75)
```
